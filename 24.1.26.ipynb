```python
#Number Pattern (Constant)
n = int(input("enter the no you want to print")) 
for i in range(1, n+1):
    for j in range(n):
        print(i, end=" ")
    print()
```

```python
#Increasing Number Pattern
x=1
n = int(input("enter the no you want to print")) 
for i in range(1, n+1):
    for j in range(i):
        print(x, end=" ")
        x += 1
    print()
```

```python
#Alphabet patterns
x=1
n = int(input("enter the no you want to print")) 
for i in range(1, n+1):
    for j in range(i):
        print(x, end=" ")
        x += 1
    print()
```

```python
#Advanced Pattern
n = 4
for i in range(1, n+1):
    print(" " * (n - i) + "* " * i)

```
