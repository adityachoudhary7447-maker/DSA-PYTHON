```python
n = int(input("Enter number of employees: "))
emp = {}

for i in range(n):
    name = input("Enter employee name: ")
    salary = int(input("Enter salary: "))
    emp[name] = salary

print(emp)
```

```python
s = input("Enter string: ")
ch = input("Enter character to count: ")

count = 0
for i in s:
    if i == ch:
        count += 1

print("Occurrence:", count)
```
