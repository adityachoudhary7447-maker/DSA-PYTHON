```python
#--Palindrome string
s = input("Enter a string")

if s == s[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
n = int(input("Enter n: "))
a, b = 0, 1

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

```python
n = int(input("Enter number: "))
fact = 1

for i in range(1, n + 1):
    fact = fact * i

print(fact)

```

```python
s = input("Enter string: ")

v = c = d = sp = 0

for ch in s:
    if ch.isdigit():
        d += 1
    elif ch.isalpha():
        if ch.lower() in "aeiou":
            v += 1
        else:
            c += 1
    else:
        sp += 1

print("Vowels:", v)
print("Consonants:", c)
print("Digits:", d)
print("Special characters:", sp)
```

```python
s = input("Enter string: ")

result = s[0]

for i in range(1, len(s)):
    if s[i] != s[i-1]:
        result += s[i]
        

print(result)
```

```python
s = input("Enter string: ")
old = input("Enter character to replace: ")
new = input("Enter new character: ")

print(s.replace(old, new))
```

```python
s = input("Enter string: ")

words = s.split()
words.reverse()

print(" ".join(words))

```

```python
s = input("Enter string: ")

res = ""
count = 1

for i in range(1, len(s)):
    if s[i] == s[i-1]:
        count += 1
    else:
        res += s[i-1] + str(count)
        count = 1

res += s[-1] + str(count)

print(res)
```
